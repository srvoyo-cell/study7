## Часть 1. Базовые операции с сигналами

### Задание 1.1. Генерация тестового сигнала

Сгенерировать сигнал, состоящий из суммы двух гармоник:

$$
x(t) = \sin(2\pi \cdot 50 \cdot t) + 0.7 \cdot \sin(2\pi \cdot 120 \cdot t)
$$

Параметры сигнала:
- Длительность: 1 секунда
- Частота дискретизации (для моделирования непрерывного сигнала): $f_{s\_orig} = 10000$ Гц

```python
# Импорт необходимых библиотек
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Параметры сигнала
f1 = 50          # Гц
f2 = 120         # Гц
A1 = 1.0
A2 = 0.7
T = 1.0          # секунд
fs_orig = 10000  # Гц
```

```python
# Временная сетка
t = np.linspace(0, T, int(fs_orig * T), endpoint=False)

# Генерация сигнала
x = A1 * np.sin(2 * np.pi * f1 * t) + A2 * np.sin(2 * np.pi * f2 * t)

# Визуализация первых 0.05 секунды
plt.figure(figsize=(12, 4))
plt.plot(t[:500], x[:500])
#plt.plot(t, x)
plt.title('Тестовый сигнал: x(t) = sin(2π·50·t) + 0.7·sin(2π·120·t)')
plt.xlabel('Время, с')
plt.ylabel('Амплитуда')
plt.grid()
```

```python
from numpy.fft import fft, fftfreq
```

```python
# БПФ
N = len(x)
X = fft(x)
freqs = fftfreq(N, 1/fs_orig)
```

```python
# Спектр (только положительные частоты) - график
plt.figure(figsize=(12, 4))
plt.plot(freqs[:N//2], np.abs(X[:N//2]))
plt.xlim(0, 150)
plt.ylim(0, 1000)
plt.title('Спектр сигнала')
plt.xlabel('Частота, Гц')
plt.ylabel('Амплитуда спектра')
plt.grid()
plt.axvline(x=f1, color='r', linestyle='--', label=f'{f1} Гц')
plt.axvline(x=f2, color='g', linestyle='--', label=f'{f2} Гц')
plt.legend()
```
